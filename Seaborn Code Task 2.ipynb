```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
bacteria_data = pd.read_csv(
    "https://raw.githubusercontent.com/HackBio-Internship/2025_project_collection/refs/heads/main/bacteria.csv"
)
```

```python
# Verify dataset loaded correctly
print(bacteria_data.shape)
bacteria_data.head()
```

```python
# Set Seaborn theme
sns.set_theme(style="whitegrid")
```

# Part 1. Distribution Plots

```python
# 1. Histogram of carb_fit
plt.figure(figsize=(8, 5))
sns.histplot(data=bacteria_data, x="carb_fit")

plt.title("Distribution of carb_fit")
plt.xlabel("carb_fit")
plt.ylabel("Count")

plt.show()

# This histogram shows the distribution of carb_fit values across all bacterial samples.
```

```python
# 2. KDE plot of cipro_fit
plt.figure(figsize=(8,5))
sns.kdeplot(data=bacteria_data, x="cipro_fit", fill=True)
plt.title("Distribution of cipro_fit")
plt.xlabel("cipro_fit")
plt.ylabel("Density")

plt.show()
```

```python
# 3. Displot showing the distribution of a fitness-related variable, grouped by Phenotype
plot = sns.displot(
    data=bacteria_data,
    x="carb_fit",
    hue="Phenotype",
    col="Phenotype",
    kind="hist",
    height=5,
    aspect=1.2
)

plot.set_axis_labels("carb_fit", "Count")
plot.fig.suptitle(
    "Distribution of carb_fit by Phenotype",
    y=1.05
)

plt.show()

# This displot shows the distribution of carb_fit values grouped by Phenotype using both color and faceting.
```

# Part 3: Categorical Plots

```python
# 1. Bar plot grouped by labels
plt.figure(figsize=(8, 5))

sns.barplot(
    data=bacteria_data,
    x="labels",
    y="carb_fit"
)

plt.title("Mean carb_fit by Labels")
plt.xlabel("Labels")
plt.ylabel("Mean carb_fit")
plt.xticks(rotation=45)

plt.show()
```

```python
# This plot compares the mean carb_fit values across different label groups.
```

```python
# 2. Box plot grouped by Phenotype
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=bacteria_data,
    x="Phenotype",
    y="carb_fit"
)

plt.title("Distribution of carb_fit by Phenotype")
plt.xlabel("Phenotype")
plt.ylabel("carb_fit")
plt.xticks(rotation=45)

plt.show()
```

```python
# The box shows the middle portion of the data, while the whiskers and points help reveal spread and potential outliers.
```

```python
# 3. Violin plot grouped by Isolation origin
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=bacteria_data,
    x="Isolation origin",
    y="carb_fit"
)

plt.title("Distribution of carb_fit by Isolation Origin")
plt.xlabel("Isolation Origin")
plt.ylabel("carb_fit")
plt.xticks(rotation=45, ha="right")

plt.show()
```

```python
# The violin shape shows the distribution/density of carb_fit within each isolation origin.
```

```python
# 4. Strip plot showing the raw observations
plt.figure(figsize=(8, 5))

sns.stripplot(
    data=bacteria_data,
    x="Phenotype",
    y="carb_fit",
    jitter=True
)

plt.title("Raw carb_fit Observations by Phenotype")
plt.xlabel("Phenotype")
plt.ylabel("carb_fit")
plt.xticks(rotation=45)

plt.show()
```

```python
# This plot displays the individual carb_fit observations for each phenotype.
```

# Part 4: Heatmap of Fitness Values

```python
# Select all columns whose names end with "_fit"
fitness_cols = ["carb_fit", "cipro_fit", "genta_fit", "kan_fit", "pip_fit"]

# Create a figure with a larger size for better readability
plt.figure(figsize=(12, 12))

# Plot the heatmap
# data: Numerical matrix of fitness values across bacterial samples
# annot=True: Displays the exact numeric value inside each cell
# fmt=".2f": Formats numerical values to two decimal places
# annot_kws: Adjusts the font size of the inside text so it fits the cell
# cmap: Default Seaborn sequential palette (rocket).
# High Fitness Values (1.2 to 1.8+): Represented by bright peach/white tones
# Medium Fitness Values (0.6 to 1.0): Represented by bright red/pink tones
# Low Fitness Values (0.0 to 0.4): Represented by dark purple / blackish-violet tones
sns.heatmap(
    bacteria_data[fitness_cols],
    annot=True,
    fmt=".2f",
    annot_kws={"size": 7},
    cmap="rocket", 
)

# Add descriptive title and axis labels
plt.title("Heatmap of Fitness Values")
plt.xlabel("Fitness Measurements")
plt.ylabel("Samples")

# Display the plot
plt.show()
```

```python
# This heatmap compares fitness measurements across samples, with numerical values shown in each cell.
```

# Part 5: Faceted Categorical Plot with catplot()

```python
# Create a faceted box plot of Expression by Gene, split by Condition
plot = sns.catplot(
    data=bacteria_data,
    x="Gene",
    y="Expression",
    col="Condition",
    kind="box",
    height=5,
    aspect=1
)

plot.set_axis_labels("Gene", "Expression")
plot.set_titles("{col_name}")
plot.fig.suptitle(
    "Expression by Gene and Condition",
    y=1.05
)

plt.show()
```
